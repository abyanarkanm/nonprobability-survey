```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sns
import numpy as np

print("✅ Semua library berhasil diimport")
```

## 📂 2. Load & Persiapan Data

```python
# Load dataset
df_raw = pd.read_csv('Form_Responses_1.csv')
print(f"Jumlah baris (responden): {len(df_raw)}")
print(f"Jumlah kolom: {df_raw.shape[1]}")
df_raw.head()
```

```python
# Drop kolom yang tidak digunakan (Program Studi, Semester)
df = df_raw.drop(columns=['Timestamp', 'Nama', 'Program Studi', 'Semester'])

# Rename kolom pertanyaan menjadi Q1-Q10
q_cols = [c for c in df.columns if c not in ['Jenis Kelamin', 'Umur']]
q_labels = [f'Q{i+1}' for i in range(len(q_cols))]
df = df.rename(columns=dict(zip(q_cols, q_labels)))

# Keterangan pertanyaan
pertanyaan = {
    'Q1':  'Menggunakan platform music streaming saat belajar',
    'Q2':  'Musik membantu lebih fokus mengerjakan tugas',
    'Q3':  'Lebih semangat belajar ditemani musik',
    'Q4':  'Musik mengurangi rasa bosan saat belajar',
    'Q5':  'Lebih cepat menyelesaikan tugas dengan musik',
    'Q6':  'Musik tidak mengganggu konsentrasi',
    'Q7':  'Produktivitas belajar meningkat dengan musik',
    'Q8':  'Lebih nyaman belajar dengan musik daripada sunyi',
    'Q9':  'Aktif memilih playlist untuk belajar',
    'Q10': 'Music streaming berdampak positif terhadap produktivitas'
}

print("✅ Data siap dianalisis")
print(f"\nKolom: {df.columns.tolist()}")
df.head()
```

## 🎯 4. Naive Estimation
Menghitung proporsi responden yang **"Setuju"** (skor ≥ 4) secara langsung tanpa koreksi:

$$\hat{P} = \frac{\text{Jumlah Setuju}}{\text{Total Responden}}$$


```python
# Naive Estimation
naive = {}
for q in q_labels:
    setuju = (df[q] >= 4).sum()
    total = len(df)
    naive[q] = setuju / total

# Tampilkan hasil
naive_df = pd.DataFrame({
    'Pertanyaan': [pertanyaan[q] for q in q_labels],
    'Jumlah Setuju (≥4)': [(df[q] >= 4).sum() for q in q_labels],
    'Total Responden': len(df),
    'Proporsi Naive': [f"{v:.4f}" for v in naive.values()],
    'Persentase (%)': [f"{v*100:.2f}%" for v in naive.values()]
}, index=q_labels)

print("=== HASIL NAIVE ESTIMATION ===")
naive_df
```

## ⚖️ 5. Weighted Estimation
Koreksi bias sampling dengan membandingkan proporsi populasi vs sampel:

$$w_i = \frac{\text{Proporsi Populasi}_i}{\text{Proporsi Sampel}_i}$$

**Data Populasi Mahasiswa Statistika UNRAM:**
- Perempuan: 126 mahasiswa
- Laki-laki: 31 mahasiswa
- **Total: 157 mahasiswa**


```python
# Data populasi
pop_total = 157
pop_P = 126 / pop_total  # Proporsi perempuan populasi
pop_L = 31 / pop_total   # Proporsi laki-laki populasi

# Proporsi sampel
n = len(df)
n_P = (df['Jenis Kelamin'] == 'Perempuan').sum()
n_L = (df['Jenis Kelamin'] == 'Laki-laki').sum()
samp_P = n_P / n
samp_L = n_L / n

# Hitung bobot
w_P = pop_P / samp_P
w_L = pop_L / samp_L

print("=" * 45)
print(f"{'':20} {'Populasi':>10} {'Sampel':>10} {'Bobot':>8}")
print("=" * 45)
print(f"{'Perempuan':20} {pop_P:>10.4f} {samp_P:>10.4f} {w_P:>8.4f}")
print(f"{'Laki-laki':20} {pop_L:>10.4f} {samp_L:>10.4f} {w_L:>8.4f}")
print("=" * 45)
print(f"\nInterpretasi:")
print(f"  - Bobot Perempuan ({w_P:.4f}): sedikit dinaikkan bobotnya")
print(f"  - Bobot Laki-laki ({w_L:.4f}): diturunkan bobotnya (over-represented)")
```

```python
# Terapkan bobot ke data
df['weight'] = df['Jenis Kelamin'].map({'Perempuan': w_P, 'Laki-laki': w_L})

# Weighted Estimation
weighted = {}
for q in q_labels:
    df['setuju_q'] = (df[q] >= 4).astype(int)
    w_setuju = (df['setuju_q'] * df['weight']).sum()
    w_total = df['weight'].sum()
    weighted[q] = w_setuju / w_total

# Tampilkan hasil
weighted_df = pd.DataFrame({
    'Pertanyaan': [pertanyaan[q] for q in q_labels],
    'Proporsi Weighted': [f"{v:.4f}" for v in weighted.values()],
    'Persentase (%)': [f"{v*100:.2f}%" for v in weighted.values()]
}, index=q_labels)

print("=== HASIL WEIGHTED ESTIMATION ===")
weighted_df
```

## 📈 6. Perbandingan Naive vs Weighted Estimation

```python
# Tabel perbandingan lengkap
comparison = pd.DataFrame({
    'Pertanyaan': [pertanyaan[q] for q in q_labels],
    'Naive (%)': [f"{naive[q]*100:.2f}%" for q in q_labels],
    'Weighted (%)': [f"{weighted[q]*100:.2f}%" for q in q_labels],
    'Selisih (pp)': [f"{(weighted[q]-naive[q])*100:+.2f}" for q in q_labels]
}, index=q_labels)

print("=== TABEL PERBANDINGAN ESTIMASI ===")
comparison
```

```python
# Grafik perbandingan
x = np.arange(len(q_labels))
width = 0.35

fig, ax = plt.subplots(figsize=(14, 6))
fig.patch.set_facecolor('#121212')
ax.set_facecolor('#1E1E1E')

b1 = ax.bar(x - width/2, [naive[q] for q in q_labels], width, label='Naive Estimation', color='#535353')
b2 = ax.bar(x + width/2, [weighted[q] for q in q_labels], width, label='Weighted Estimation', color='#1DB954')

ax.set_xticks(x); ax.set_xticklabels(q_labels, color='white', fontsize=11)
ax.set_ylabel("Proporsi 'Setuju' (Skor ≥ 4)", color='white', fontsize=11)
ax.set_ylim(0, 1.1)
ax.set_title("Perbandingan Naive vs Weighted Estimation", color='white', fontsize=13, fontweight='bold', pad=15)
ax.tick_params(colors='white')
for sp in ['top','right']: ax.spines[sp].set_visible(False)
for sp in ['bottom','left']: ax.spines[sp].set_color('#444')
ax.axhline(0.5, color='white', linestyle=':', linewidth=1, alpha=0.4)
ax.legend(facecolor='#2A2A2A', labelcolor='white', fontsize=11)
plt.tight_layout()
plt.savefig('grafik3_naive_vs_weighted.png', dpi=150, bbox_inches='tight', facecolor='#121212')
plt.show()
```

## 📝 7. Kesimpulan

### Temuan Utama:
1. **Responden** berjumlah **28 mahasiswa** Statistika UNRAM (21 perempuan, 7 laki-laki)
2. **Q4 (Musik mengurangi rasa bosan)** memiliki proporsi setuju tertinggi: **~89.5%**
3. **Q5 (Lebih cepat menyelesaikan tugas)** memiliki proporsi setuju terendah: **~40.9%**
4. **Perbedaan Naive vs Weighted** relatif kecil (selisih < 2pp), karena bias sampling tidak terlalu besar

### Catatan Metodologi:
- Sampling menggunakan **convenience sampling** (non-probability)
- Laki-laki **over-represented** di sampel (25%) dibanding populasi (19.75%)
- Weighted estimation memberikan gambaran yang **lebih representatif** terhadap populasi sesungguhnya
